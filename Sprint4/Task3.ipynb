```
using System.Runtime.Intrinsics.X86;

    enum Platform
    {
        IOS,
        ANDROID,
        PC
    }

    interface IDisplayNotif
    {
        void DisplayNotification();
    }

    interface ISendNotif
    {
        void SendNotification(string recipient);
    }

    interface IGotNotif
    {
        void GetNotificationDetails();
    }

    interface IStrange
    {
        void StrangeFunction();
    }

    abstract class Notification : IDisplayNotif, IGotNotif
    {
        int _notificationid;
        public string _messageText
        {
            get;
            protected set;
        }
        
        public Notification(string messageText)
        {
            _notificationid = new Random().Next(0, 999999);
            _messageText = messageText;
        }

        public virtual void DisplayNotification()
        { 
            Console.WriteLine(_messageText);
        }

        public virtual void GetNotificationDetails()
        {
            Console.WriteLine("notif id: ", _notificationid.ToString(), ";");
            Console.WriteLine("message: ", _messageText);
        }
        
    };

    class EmailNotification : Notification, ISendNotif
    { 
        public string _emailAdress
        {
            get;
            private set;
        }

        public EmailNotification(string messageText, string email) : base(messageText)
        {
            _emailAdress = email;
        }

        public override void DisplayNotification()
        { 
            Console.WriteLine(_emailAdress, ": ", _messageText);
        }

        public void SendNotification(string recipient)
        {
            Console.WriteLine("Notification sent from ", _emailAdress, " to ", recipient);
        }
    };

    class SMSNotification : Notification, ISendNotif
    {
        public string _phoneNumber
        {
            get;
            private set;
        }

        public SMSNotification(string messageText, string phoneNumber) : base(messageText) 
        { 
            _phoneNumber = phoneNumber;
        }

        public void SendNotification(string recipient)
        {
            Console.WriteLine("Notification sent from ", _phoneNumber, " to ", recipient);
        }
    };

    class PushNotification : Notification, ISendNotif, IStrange
    {
        public Platform _platform
        {
            get;
            private set;
        }

        public PushNotification(string messageText, Platform platform) : base(messageText) 
        {
            _platform = platform;
        }

        public void SendNotification(string recipient)
        {
            if(_platform == Platform.IOS)
                Console.WriteLine("Notification sent to ", recipient, " platrofotm: IOS");
            if (_platform == Platform.ANDROID)
                Console.WriteLine("Notification sent to ", recipient, " platrofotm: ANDROID");
            if (_platform == Platform.PC)
                Console.WriteLine("Notification sent to ", recipient, " platrofotm: PC");
        }

        public void StrangeFunction()
        {
            Console.WriteLine("wow");
        }
    };

    class NotifCollector<T> where T : Notification
    {
        private List<T> _notifs = new List<T>();

        public void Add(T notif)
        {
            _notifs.Add(notif);
        }

        public void Remove(T notif)
        {
            _notifs.Remove(notif);
        }

        public void DisplayAnimals()
        {
            foreach (var notif in _notifs)
            {
                notif.DisplayNotification(); 
            }
        }
    }

    class ScaryNotif : Notification, ISendNotif
    {
        private readonly IDisplayNotif _display;

        ScaryNotif(string messageText, IDisplayNotif display) : base(messageText) 
        {
            _display = display;
        }

        void ISendNotif.SendNotification(string recipient)
        {
            _display.DisplayNotification();
        }// переопределить оператор + 
        // метод множественного возвращение с рефлексией c#

    }
```
